```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tax = pd.read_csv('./data/Tax_info.csv')
onlinesales = pd.read_csv('./data/Onlinesales_info.csv')
marketinginfo = pd.read_csv('./data/Marketing_info.csv')
discountinfo = pd.read_csv('./data/Discount_info.csv')
customerinfo = pd.read_csv('./data/Customer_info.csv')
```

```python
dfs = {
    "tax": tax,
    "onlinesales": onlinesales,
    "marketinginfo": marketinginfo,
    "discountinfo": discountinfo,
    "customerinfo": customerinfo,
}

for name, df in dfs.items():
    print("="*80)
    print(f"[{name}] shape: {df.shape}")
    display(df.head())
```

```python
for name, df in dfs.items():
    print("="*80)
    print(f"[{name}] info()")
    df.info()
```

```python
for name, df in dfs.items():
    print("="*80)
    print(f"[{name}] 결측치 요약")

    na_cnt = df.isna().sum()
    na_ratio = na_cnt / len(df)

    summary = pd.DataFrame({
        "na_count": na_cnt,
        "na_ratio": na_ratio
    }).sort_values("na_ratio", ascending=False)

    display(summary[summary["na_count"] > 0])
```

```python
for name, df in dfs.items():
    dup_cnt = df.duplicated().sum()
    print(f"[{name}] duplicated rows: {dup_cnt}")
```

```python
for name, df in dfs.items():
    print("="*80)
    print(f"[{name}] numeric summary")

    num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    print("numeric columns:", num_cols)

    if len(num_cols) == 0:
        print("숫자형 컬럼 없음")
        continue

    display(df[num_cols].describe().T)
```

```python
TARGET = "onlinesales"

if TARGET not in dfs:
    print(f"⚠️ '{TARGET}' 데이터가 없습니다.")
else:
    df = dfs[TARGET]
    num_cols = df.select_dtypes(include=[np.number]).columns.tolist()

    for col in num_cols[:3]:
        plt.figure(figsize=(6,3))
        plt.hist(df[col].dropna(), bins=50)
        plt.title(f"{TARGET} :: {col}")
        plt.tight_layout()
        plt.show()
```
